```python
%sql
INSERT INTO workspace.silver.dim_date
-- Registro centinela para claves huérfanas o desconocidas (DateKey = -1)
SELECT 
    -1 AS DateKey,
    to_date('1900-01-01') AS FullDate,
    -1 AS Year,
    -1 AS Quarter,
    -1 AS Month,
    'Unknown' AS MonthName,
    -1 AS Day,
    'Unknown' AS DayOfWeekName

UNION ALL

-- Rango continuo de fechas desde 2020 hasta 2030
SELECT 
    CAST(date_format(cal_date, 'yyyyMMdd') AS INT) AS DateKey,
    cal_date AS FullDate,
    year(cal_date) AS Year,
    quarter(cal_date) AS Quarter,
    month(cal_date) AS Month,
    date_format(cal_date, 'MMMM') AS MonthName,
    day(cal_date) AS Day,
    date_format(cal_date, 'EEEE') AS DayOfWeekName
FROM (
    SELECT explode(sequence(to_date('2020-01-01'), to_date('2030-12-31'), interval 1 day)) AS cal_date
);
```
