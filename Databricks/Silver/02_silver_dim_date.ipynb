```python
%sql
-- Full Calendar Population (2015 to 2035) + Sentinel -1
INSERT INTO workspace.silver.dim_date (
    DateKey,
    FullDate,
    YearNbr,
    QuarterNbr,
    MonthNbr,
    MonthName,
    DayNbr,
    DayOfWeekName
)
-- Sentinel Record (Kimball Standard: -1)
SELECT 
    -1 AS DateKey,
    DATE('1900-01-01') AS FullDate,
    -1 AS YearNbr,
    -1 AS QuarterNbr,
    -1 AS MonthNbr,
    'Unknown' AS MonthName,
    -1 AS DayNbr,
    'Unknown' AS DayOfWeekName

UNION ALL

-- Daily calendar date generation (2015-01-01 to 2035-12-31)
SELECT 
    CAST(date_format(d, 'yyyyMMdd') AS INT) AS DateKey,
    d AS FullDate,
    year(d) AS YearNbr,
    quarter(d) AS QuarterNbr,
    month(d) AS MonthNbr,
    date_format(d, 'MMMM') AS MonthName,
    day(d) AS DayNbr,
    date_format(d, 'EEEE') AS DayOfWeekName
FROM (
    SELECT explode(sequence(DATE('2015-01-01'), DATE('2035-12-31'), INTERVAL 1 DAY)) AS d
);
```
